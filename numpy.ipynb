```python
# Create a 5*5 matrix of random integers (1-90)

mat = ap.random.randint(1,90,(5,5))
print(mat)
```

```python
# Find mean of vales row and column wise.
mat = ap.random.randint(1,90,(5,5))
print(mat)
print(ap.mean(mat,axis=0))
print(ap.mean(mat,axis=1))
```

```python
# Replace all even numbers with -1
mat = ap.random.randint(1,90,(5,5))
mat1 = ap.where(mat%2==0,-1,mat)
print(mat1)
```

```python
# sort the array row and column wise.
mat = ap.random.randint(1,90,(5,5))
mat = ap.sort(mat,axis=0)
mat = ap.sort(mat,axis=1)
print(mat)
```

```python
# Find row and column wise sum separately.
mat = ap.random.randint(1,90,(5,5))
mat = ap.sum(mat,axis=0)
mat = ap.sum(mat,axis=0)
print(mat)
```

```python
# count the number of -ve in each column
mat = ap.random.randint(-90,90,(5,5))
mat = ap.where(mat<0,1,0)
mat = ap.sum(mat,axis=0)
print(mat)
```

```python
# inventory control system
stock = ap.random.randint(1,20,(5,5))
demand = ap.random.randint(1,20,(5,5))
remaining_stock = stock - demand
print(remaining_stock)

modified = ap.where(remaining_stock<0,0,remaining_stock)
print(modified)

total_shortage = ap.sum(ap.where(remaining_stock<0,remaining_stock  -1,0))
print(total_shortage)

product_shortage = ap.sum(ap.where(remaining_stock<0,1,0))
print(total_shortage,product_shortage)

sorted = ap.sort(modified,axis = 1)

```
